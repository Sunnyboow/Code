```python

users = [
    {'name':'James','age':31},
    {'name':'Mike','age':32},
    {'name':'Emmy','age':45},
    {'name':'Antony','age':27}
    
]



new_list = []
for i in users:
    new_list.append(i['age'])
    
max(new_list) 
```
